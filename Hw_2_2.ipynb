```python
file_path = '/content/drive/My Drive/2017_jun_final.csv'

df = pd.read_csv(file_path)

df
```

```python
# частка пропусків в колонці в %
df.isnull().sum()/(df.shape[0]-1)*100
```

```python
#видаляємо всі стовпці з пропусками крім стовпця "Мова програмування"
df1=df['Мова програмування']
df2=df.dropna(axis=1)
new_df=pd.concat([df1,df2],axis=1)

```

```python
# частка пропусків в колонці в %
new_df.isnull().sum()/(new_df.shape[0]-1)*100
```

```python
new_df.shape
```

```python
new_df=df.groupby('Посада')['Зарплата на місяць'].agg(['min','max'])
new_df
```

```python
def fill(val):
  return val.mean()
df1=df.groupby('Посада')['Зарплата на місяць'].apply(fill)
new_df['avg']=df1
new_df
```

```python
new_df.describe()
```

```python
new_df.to_csv('salary.csv',index=False)
```

```python
fig,axs=plt.subplots()
axs.plot(new_df['max'],label='max')
axs.plot(new_df['avg'],label='mean')
axs.plot(new_df['min'],label='min')
plt.title('Зарплати на посадах ')
plt.legend()
plt.show()
```

```python
#гістограмма максимальної зарплати'
data=new_df['max']
plt.hist(data,bins=10)
plt.show
```
