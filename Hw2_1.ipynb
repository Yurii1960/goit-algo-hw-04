```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

import requests
from io import StringIO

url = "https://uk.wikipedia.org/wiki/%D0%9D%D0%B0%D1%81%D0%B5%D0%BB%D0%B5%D0%BD%D0%BD%D1%8F_%D0%A3%D0%BA%D1%80%D0%B0%D1%97%D0%BD%D0%B8#%D0%9D%D0%B0%D1%80%D0%BE%D0%B4%D0%B6%D1%83%D0%B2%D0%B0%D0%BD%D1%96%D1%81%D1%82%D1%8C"

headers = {
"User-Agent": "Mozilla/5.0",
"Accept-Language": "uk-UA,uk;q=0.9,en;q=0.8",
}

html = requests.get(url, headers=headers, timeout=30).text

df = pd.read_html(
StringIO(html),
match="Коефіцієнт народжуваності в регіонах України",
thousands=".", decimal=","
)[0]

df
```

```python
df.head(3)
```

```python
df.shape
```

```python
df=df.replace('—',np.nan)
df
```

```python
df.dtypes
```

```python
df['2014']=pd.to_numeric(df['2014'])
df['2019']=pd.to_numeric(df['2019'])
df.dtypes
```

```python
# частка пропусків в колонці в %
df.isnull().sum()/(df.shape[0]-1)*100
```

```python
#видаляємо останній рядок
df=df[:-1]
df
```

```python
#заміняємо відсутні значення середніми
new_df=df.drop(['Регіон'],axis=1)
columns_name=new_df.columns

for col in columns_name:
  df=df.fillna({col:new_df[col].mean()})
df
```

```python
#список регіонів де народжуваність у 2019 році вища за середню по країні
df.loc[df['2019']>df['2019'].mean()]['Регіон']
```

```python
#регіон з найбільщою народжуваністю
df[df['2014']==df['2014'].max()]['Регіон']
```

```python
import matplotlib.pyplot as plt
df1=df['Регіон']
df2=df['2019']
new_df=pd.concat([df1,df2],axis=1)
new_df.plot.bar()
plt.title('Коефіцієнти народжуваності по регіонах у 2019 році')
plt.show()


```

```python
fig,axs=plt.subplots()
data_k=df.loc[25][1:]
data_z=df.loc[6][1:]
data_x=df.loc[20][1:]
axs.plot(data_k,label='Київ')
axs.plot(data_z,label='Закарпатська')
axs.plot(data_x,label='Херсонська')
plt.title('Коефіцієнти народжуваності по роках ')
plt.legend()
plt.show()
```
